```python
import numpy as np
import matplotlib.pyplot as plt
distance=[0,1,2,3,4,5]
BOOKING_FEE=50
def delivery_charge(distance):
    return BOOKING_FEE+15*distance
result=delivery_charge(5)
print(result)


    
```

```python
charges = [delivery_charge(d) for d in distance]
plt.plot(distance,charges)
plt.xlabel("distance")
plt.ylabel("charges")
plt.title("DISTANCE CHARGES")
```
