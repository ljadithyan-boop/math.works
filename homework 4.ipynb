```python
import numpy as np
student_scores = np.array([  [75, 82, 88],  [68, 79, 91],   [80, 85, 95]])
print(student_scores)
print("THE SHAPE OF THE MATRIX IS:",student_scores)
```

```python

improvement_scores = np.array([ [5, 4, 3],[6, 5, 4],  [4, 3, 2]])
result=(student_scores+improvement_scores)
print("THE ADDITION OF 2 MATRICES IS:",result)
```

```python
result1=(2*student_scores)
print("THE RESULT AFTER MULTIPLICATION IS:",result1)
```

```python
repo=student_scores.T
print("THE TRANSPOSE OF MATRIX IS:",repo)
```

```python
B = np.array([[2, 1, 3],[1, 4, 2], [3, 2, 5]])
fe=np.linalg.det(B)
print("THE DETERMINANT OF ARRAY IS:",fe)
if fe==0:
    print("it is a singular matrix")
else:
    print("it is a non singular matrix")
abc=np.linalg.inv(B)
print("THE INVERSE OF THE MATRIX IS:",abc)
```

```python
s=np.eye(4)
print("THE 4*4 MATRIX IS:",s)
```

```python
sop=np.linalg.matrix_rank(B)
print("THE RANK OF B MATRIX IS:",sop)
```
