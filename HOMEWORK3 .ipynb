```python
import numpy as np
location_a=np.array([3,4])
location_b=np.array([5,7])
print("THE OG:",location_a)
print("THE OG:",location_b)
```

```python
result=location_a+location_b
print("THE ADDED VECTOR IS:",result)
```

```python
point_a=4*location_a
point_b=2*location_b
print("MULTIPLIED BY 4 :",point_a)
print("MULTIPLIED BY 2:",point_b)
```

```python
magnitude=np.linalg.norm(location_a)
magnitude1=np.linalg.norm(location_b)
print("THE MAGNITUDE OF POINT A IS:",magnitude)
print("THE MAGNITUDE OF POINT B IS:",magnitude1)

```
