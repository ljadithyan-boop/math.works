```python
import numpy as np
import matplotlib.pyplot as plt
REGISTRATION_FEE=500
months=[0,1,2,3,4,5,6]
def membership_cost(months):
    return 500+300*months
costs = []
for m in months:
    cost = membership_cost(m)
    costs.append(cost)
    print(f"Month {m}: {cost}")
plt.plot(months,costs)
plt.xlabel("months")
plt.ylabel("costs")
plt.title("MONTHS VS COST")
```
