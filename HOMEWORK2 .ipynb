```python
import statistics
import numpy as np
sales = [12, 15, 18, 20, 15, 22, 25, 15, 28, 30]
sales.sort()
median=statistics.median(sales)
print("THE MEDIAN IS:",median)
mean=sum(sales)/len(sales)
print("THE MEAN IS:",mean)
mode=statistics.mode(sales)
print("THE MODE IS :",mode)

```
