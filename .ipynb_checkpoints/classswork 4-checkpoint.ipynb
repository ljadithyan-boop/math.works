```python
import numpy as np
sales_data = np.array([   [120, 150, 180],
                          [100, 130, 170], 
                          [140, 160, 190]])
print(sales_data)
print("THE SHAPE OF THE DATA IS:",sales_data.shape)
```

```python
bonus_sales = np.array([ [10, 15, 20], [5, 10, 15],  [12, 14, 18]])
result=sales_data+bonus_sales
print("THE ADDITION OF BOTH THE MATRICES:",result)
```

```python
RESULT1=2*sales_data
print("THE MULTIPLICATION OF THE OG MATRIX:",RESULT1)
```

```python
senop=sales_data.T
print("THE TEANSPOSE OF MATRIX IS:",senop)
```

```python
new_matrix = np.array([ [4, 2, 1], [3, 5, 2],[1, 2, 3]])
se=np.linalg.det(new_matrix)
print("THE DETERMINANT OF NEW MATRIX IS:",se)
if se==0:
    print("singular_matrix")
else:
    print("non singular_matrix")
fe=np.linalg.inv(new_matrix)
print("THE INVERSE OF THE MATRIX IS:",fe)
```

```python
rank=np.linalg.matrix_rank(new_matrix)
print("THE RANK OF MATRIX IS:",rank)
```
